**Exercício 05: Tabuada**

**Lógica:** Primeiro eu peço para o usuário digitar um número. Em sequência, eu uso a estrutura de repetição while pra calcular a tabela de 1 a 10. Com a variável contador, que começa em 1, vai aumentando em unidade a cada repetição, até chegar no valor 10.

```python
numero = int(input("Tabuada do número: "))
contador = 1
while contador <= 10:
  print(f"{numero} x {contador} = {numero * contador}")
  contador = contador + 1
```

**Exercício 06: Números Pares**

**Lógica:** Aqui, pensei na lógica dos números pares. Sempre o resto da divisão por 2 deve ser igual a 0. Então coloquei na estrutura de repetição while, com o contador, para percorrer os números de 1 até 20. Dentro da repetição, usei if para verificar se o número era par. Quando é verdadeiro, o número é exibido.

```python
print ("Números pares de 1 a 20: ")
contador = 1

while contador <= 20:
  if contador % 2 == 0:
    print(contador)
  contador = contador + 1
```

**Exercício 07: Números ímpar ou par**

**Lógica:**  Utilizei a regra dos números pares, em que o resto da divisão por 2 deve ser igual a 0. Para verificar isso, usei a estrutura condicional if. Se o número digitado pelo usuário tiver resto 0 na divisão por 2, ele será par. Caso contrário, será ímpar.

```python
numero = int(input("Digite um número para verificar se é PAR ou ÍMPAR: "))

if numero % 2 == 0:
  print(f"{numero} é par!")
else:
  print(f"{numero} é ímpar!")
```

**Exercício 08: Verificação de Palíndromo**

**Lógica:** Comecei pensando que, para descobrir se uma palavra era um palíndromo, seria necessário comparar com ela mesma, porém escrita ao contrário. Para isso utilizei [::-1], que inverte a ordem dos caracteres da palavra, então ele pega a palavra inteira, mas caminha de trás pra frente. Em sequência, usei a estrutura condicional if para comparar a palavra original com a palavra invertida. Se ambas fossem iguais, a palavra era um palíndromo, caso contrário, não era.

```python
palavra = input("Digite uma palavra: ")

if palavra == palavra [::-1]:
  print("A palavra é um palíndromo!")
else:
  print("A palavra não é um palíndromo!")
```

**Exercício 09: Fatorial de um número**

**Lógica:** Solicitei que o usuário digitasse um número e criei uma variável com o valor inicial 1, porque o cálculo do fatorial é feito por multiplicação. Depois, usei a estrutura de repetição while para multiplicar o resultado pelo número digitado e diminuir esse número de 1 em 1 a cada repetição, até chegar a 1.

```python
numero = int(input("Digite um número: "))
resultado = 1

while numero > 1:
  resultado = resultado * numero
  numero = numero - 1

print(resultado)
```

**Exercício 10: Sequência de Fibonacci**

Lógica: A lógica de Fibonacci é que o próximo número seja a soma dos dois anteriores. Primeiro, defini duas variáveis com os valores iniciais 0 e 1 e uma terceira variável para controlar a quantidade de números exibidos. Depois, utilizei a estrutura de repetição while para continuar até completar 10 números. A cada repetição, ele mostra o primeiro valor, calcula o próximo número somando o primeiro com o segundo e em seguida, atualiza as variáveis para continuar a sequência.

```python
primeiro = 0
segundo = 1
contador = 0

while contador < 10:
  print(primeiro)
  proximo = primeiro + segundo
  primeiro = segundo
  segundo = proximo

  contador = contador + 1

```
